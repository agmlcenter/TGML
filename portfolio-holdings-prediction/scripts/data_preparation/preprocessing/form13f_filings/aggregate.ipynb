```python
import pandas as pd

df = pd.read_csv('sp500_final_common.csv', dtype={'FIGI': str})
duplicates = df[df.duplicated(subset=['CIK', 'CUSIP', 'source_file'], keep=False)]
#print(duplicates)

# Aggregate duplicates
# Sum numeric columns, take first for others
numeric_cols = ['VALUE', 'SSHPRNAMT']
agg_dict = {col: 'sum' if col in numeric_cols else 'first' for col in df.columns}
df_cleaned = df.groupby(['CIK', 'CUSIP', 'source_file'], as_index=False).agg(agg_dict)
# Verify no duplicates remain
duplicates_cleaned = df_cleaned[df_cleaned.duplicated(subset=['CIK', 'CUSIP', 'source_file'], keep=False)]
print(f"Number of duplicate rows after cleaning: {len(duplicates_cleaned)}")
# Save cleaned dataset
df_cleaned.to_csv('sp500_final_common_cleaned.csv', index=False)

# Validation
print(f"Original dataset rows: {len(df)}")
print(f"Cleaned dataset rows: {len(df_cleaned)}")
print(f"Rows removed: {len(df) - len(df_cleaned)}")
print("Saved to 'sp500_final_common_cleaned.csv'")
```
