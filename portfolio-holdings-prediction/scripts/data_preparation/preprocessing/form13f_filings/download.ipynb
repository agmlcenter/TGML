```python
from sec_edgar_downloader import Downloader

# Initialize downloader with email
dl = Downloader("emad.izadifar@gmail.com")

# Use an institutional investor's CIK (e.g., BlackRock: 0001364742)
cik = "0001364742"

# Download the most recent 13F filings
dl.get("13F-HR", cik, amount=2)  # You can increase the amount as needed
```

```python
import os
from bs4 import BeautifulSoup

def extract_cusips_from_file(file_path):
    with open(file_path, "r", encoding="utf-8") as f:
        soup = BeautifulSoup(f.read(), "lxml")

    cusips = []
    for info_table in soup.find_all("infotable"):
        cusip_tag = info_table.find("cusip")
        if cusip_tag:
            cusips.append(cusip_tag.text.strip())

    return cusips

# Path to the folder where the 13F files are downloaded
folder = os.path.join("sec-edgar-filings", "13F-HR", cik)

# Extract CUSIPs from all relevant files
all_cusips = []
for root, _, files in os.walk(folder):
    for file in files:
        if file.endswith(".xml") or file.endswith(".txt"):
            file_path = os.path.join(root, file)
            try:
                cusips = extract_cusips_from_file(file_path)
                all_cusips.extend(cusips)
            except Exception as e:
                print(f"Error reading {file}: {e}")

# Display unique CUSIPs
print("Found CUSIPs:", list(set(all_cusips)))
```
