```python
import pandas as pd

df = pd.read_csv('sp500_final_common_cleaned.csv', dtype={'FIGI': str, 'CIK': str, 'CUSIP': str})
df_time = pd.read_csv('timestamp.csv')
str_to_int = {df_time['string'].iloc[i]: df_time['integer'].iloc[i] for i in range(len(df_time))}
df_13f = df[['source_file', 'CIK', 'CUSIP', 'VALUE']]
df_13f = df_13f.rename(columns={'source_file': 'Year'})
df_13f['Year'] = df_13f['Year'].map(str_to_int)
df_13f = df_13f.sort_values(by=['Year', 'CIK', 'CUSIP'])
df_13f = df_13f[df_13f['Year'] != '2013q2_form13f.csv']
df_13f.to_csv(r'C:/Users/emadi/AppData/Local/Programs/Python/Python312/Lib/site-packages/tgb/datasets/tgbn_13f/tgbn-13f_node_labels.csv', index=False)
print("13F data saved to tgbn-13f_node_labels.csv")
```
