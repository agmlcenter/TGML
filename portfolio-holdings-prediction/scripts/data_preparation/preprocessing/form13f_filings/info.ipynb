```python
unique_tradeofclass = set()
for chunk in pd.read_csv('sp500_figi.csv', chunksize=10000000):
    unique_tradeofclass.update(chunk['TITLEOFCLASS'].unique())
    
print(f"Unique tradeofclass values: {unique_tradeofclass}")
print(f"Number of unique tradeofclass values: {len(unique_tradeofclass)}")

common_stock_list = ['Common Stock',
'COMMON',
'COM',
'Com',
'CL A',
'Cl A',
'CL a',
'CLA',
'Class A',
'CLASS A',
'SHS CLASS A',
'SHS Class A',
'CAP STK CL A',
'COM CL A',
'CL A COM',
'CL B',
'Class B',
'CLASS B',
'CL B COM',
'CL B NEW',
'CLASS B COM',
'CL C',
'Class C',
'CLASS C',
'CAP STK CL C',
'SHS CL A',
'SHS - A -',
'SHS CLASS -A -',
'COM SHS',
'Stock',
'Equity',
'EQUITY',
'COMMON STOCK',
'SH',
'SHS',
'SH BEN INT',
'SH BEN INT NEW',
'NEW SH BEN INT NEW',
'ORD',
'ORD SHS',
'COM STK',
'A',
'B',
'Common CLA'
]

filtered_edges = 0
filtered_chunks = []
for chunk in pd.read_csv('sp500_figi.csv', chunksize=10000000):
    filtered_chunk = chunk[chunk['TITLEOFCLASS'].isin(common_stock_list)]
    chunk_edges = len(chunk)
    filtered_edges += len(filtered_chunk)
    if not filtered_chunk.empty:
        filtered_chunks.append(filtered_chunk)
    print(f"Processed chunk with {chunk_edges} rows, kept {len(filtered_chunk)} S&P 500 rows")
    
if filtered_chunks:
    filtered_df = pd.concat(filtered_chunks)
    filtered_df.to_csv('sp500_figi.csv', index=False)
    print(f"Saved {filtered_edges} rows to sp500_figi.csv")
```

```python
df = pd.read_csv('sp500_figi.csv')
sp500_cusips = df['CUSIP'].unique()
total_edges = 0
filtered_edges = 0
unique_ciks = set()
unique_cusips = set()

filtered_chunks = []
for chunk in pd.read_csv('merged_dataset_upper.csv', chunksize=10000000):
    unique_ciks.update(chunk['CIK'])
    unique_cusips.update(chunk['CUSIP'])
    chunk_edges = len(chunk)
    total_edges += chunk_edges
    # Filter for S&P 500 CUSIPs
    filtered_chunk = chunk[chunk['CUSIP'].isin(sp500_cusips)]
    filtered_edges += len(filtered_chunk)
    if not filtered_chunk.empty:
        filtered_chunks.append(filtered_chunk)
    print(f"Processed chunk with {chunk_edges} rows, kept {len(filtered_chunk)} S&P 500 rows")

# Save filtered dataset
if filtered_chunks:
    filtered_df = pd.concat(filtered_chunks)
    filtered_df.to_csv('sp500.csv', index=False)
    print(f"Saved {filtered_edges} rows to sp500.csv")

# Print results
print(f"Total edges: {total_edges}")
print(f"Filtered S&P 500 edges: {filtered_edges}")
print(f"Unique CIKs: {len(unique_ciks)}")
print(f"S&P 500 CIKs: {len(filtered_df['CIK'].unique())}")
print(f"Unique CUSIPs: {len(unique_cusips)}")
print(f"S&P 500 CUSIPs: {len(filtered_df['CUSIP'].unique())}")
print(f"Unique FIGIs: {len(filtered_df['FIGI'].unique())}")
```
