```python
import pandas as pd
import glob
import os

# Option 1: Merge all CSV files in a directory
def merge_csv_in_directory(directory_path, output_file='merged_dataset.csv'):
    # Get all CSV files in the directory
    all_files = glob.glob(os.path.join(directory_path, "*.csv"))
    
    # Create a list to hold each DataFrame
    df_list = []
    
    # Read each CSV file into a DataFrame and add to the list
    for file in all_files:
        df = pd.read_csv(file)
        # Optionally add a column to identify source file
        df['source_file'] = os.path.basename(file)
        df_list.append(df)
    
    # Concatenate all DataFrames
    merged_df = pd.concat(df_list, ignore_index=True)
    
    # Save the merged DataFrame to a new CSV file
    merged_df.to_csv(output_file, index=False)
    
    return merged_df

merged_data = merge_csv_in_directory(r'E:/عماد/آموزش/یادگیری ماشین گراف/code/api/13f_data/')
```
