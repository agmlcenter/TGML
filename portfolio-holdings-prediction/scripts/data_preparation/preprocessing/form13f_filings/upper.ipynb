```python
import pandas as pd

# Read CSV in chunks
chunk_size = 10000000  # Adjust based on memory

filtered_chunks = []
for chunk in pd.read_csv('merged_dataset.csv', chunksize=chunk_size):
    chunk['CUSIP'] = chunk['CUSIP'].str.upper()
    filtered_chunks.append(chunk)
    print(f"Processed chunk with {len(chunk)} rows")
if filtered_chunks:
    filtered_df = pd.concat(filtered_chunks)
    filtered_df.to_csv('merged_dataset_upper.csv', index=False)
print(f"Saved merged_dataset_upper.csv")
```
