```python
import pandas as pd

# Initialize counters and mappings
total_edges = 0
filtered_edges = 0
unique_ciks = set()
unique_cusips = set()
unique_figis = set()

df_cusip = pd.read_csv('matching.csv', dtype={'FIGI': str})
sp500_cusip = df_cusip['CUSIP'].unique()

# Filter dataset for S&P 500 FIGIs
chunk_size = 10000000
filtered_chunks = []
for chunk in pd.read_csv('merged_dataset_upper.csv', chunksize=chunk_size, dtype={'FIGI': str}):
    unique_ciks.update(chunk['CIK'])
    unique_cusips.update(chunk['CUSIP'])
    unique_figis.update(chunk['FIGI'])
    chunk_edges = len(chunk)
    total_edges += chunk_edges
    # Filter for S&P 500 FIGIs
    filtered_chunk = chunk[chunk['CUSIP'].isin(sp500_cusip)]
    filtered_edges += len(filtered_chunk)
    if not filtered_chunk.empty:
        filtered_chunks.append(filtered_chunk)
    print(f"Processed chunk with {chunk_edges} rows, kept {len(filtered_chunk)} S&P 500 rows")

# Save filtered dataset
if filtered_chunks:
    filtered_df = pd.concat(filtered_chunks)
    filtered_df.to_csv('sp500_temporary.csv', index=False)
    print(f"Saved {filtered_edges} rows to sp500_temporary.csv")


# Print results
print(f"Total edges: {total_edges}")
print(f"S&P 500 edges: {filtered_edges}")
print(f"Unique CIKs: {len(unique_ciks)}")
print(f"S&P 500 CIKs: {len(filtered_df['CIK'].unique())}")
print(f"Unique CUSIPs: {len(unique_cusips)}")
print(f"S&P 500 CUSIPs: {len(filtered_df['CUSIP'].unique())}")
print(f"Unique FIGIs: {len(unique_figis)}")
print(f"S&P 500 FIGIs: {len(filtered_df['FIGI'].unique())}")
```
