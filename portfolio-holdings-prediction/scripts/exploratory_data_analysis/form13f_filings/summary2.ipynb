```python
import pandas as pd

#df = pd.read_csv('merged_dataset.csv', nrows=5)
#df.head()
#df.info()

# Initialize sets for unique values
unique_ciks = set()
unique_cusips = set()
total_edges = 0

# Read CSV in chunks
chunk_size = 10000000  # Adjust based on memory
for chunk in pd.read_csv('merged_dataset_upper.csv', chunksize=chunk_size):
    unique_ciks.update(chunk['CIK'])
    unique_cusips.update(chunk['CUSIP'])
    total_edges += len(chunk)
    print(f"Processed chunk with {len(chunk)} rows")

print('Number of unique CIKs:', len(unique_ciks))
print('Number of unique CUSIPs:', len(unique_cusips))
print('Number of unique nodes: ', len(unique_ciks) + len(unique_cusips))
print('Number of unique temporal edges: ', total_edges)
#print('Number of unique temporal edges: ', len(df))
#print('Number of unique CIKs: ', df['CIK'].unique())
#print('Number of unique CUSIPs: ', df['CUSIP'].unique())
#print('Number of unique nodes: ', df['CUSIP'].unique() + df['CIK'].unique())
#print('Number of unique temporal edges: ', len(df))'''
```
