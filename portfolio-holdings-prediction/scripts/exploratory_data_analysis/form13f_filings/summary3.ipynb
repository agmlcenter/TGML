```python
import pandas as pd

# Load the CSV with FIGI and CUSIP columns
df = pd.read_csv('sp500_figi.csv')

# Initialize an empty list to collect rows
rows = []

# Group by FIGI
grouped = df.groupby('FIGI')

for figi, group_data in grouped:
    print(f"\nGroup: {figi}")
    unique_cusips = group_data['CUSIP'].unique()
    print('Unique CUSIPs:', unique_cusips)
    #(group_data)
    print("-" * 40)  # Separator between groups
    
    if len(unique_cusips) > 1:
        # Create a row: [FIGI, CUSIP1, CUSIP2, ...]
        row = [figi] + list(unique_cusips)
        rows.append(row)

# Create the new DataFrame
max_cusips = max(len(row) - 1 for row in rows)  # max number of CUSIPs
columns = ['FIGI'] + [f'CUSIP_{i+1}' for i in range(max_cusips)]
df2 = pd.DataFrame(rows, columns=columns)

# Save to Excel
df2.to_excel('figi_to_cusip.xlsx', index=False)
```
